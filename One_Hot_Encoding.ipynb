1.one hot encoding is used for NOMINAL categorical data.

2.It is very important to convert string categorical data into numerical data and hence this technique is used.

3.We convert it into vector.

4.We convert each category into a new column. eg- if we have 50 categories , 50 new columns will be created and then vectorize it .

5.But due to "multicollinearity" and to avoid inter dependency between input or dummy variable trap columns we use n-1 columns .

```python
pd.get_dummies(df,columns=['fuel','owner'])
```

```python
pd.get_dummies(df,columns=['fuel','owner'],drop_first=True)
```

In ML , we use sklearn

3. one hot encoding using sklearn

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest= train_test_split(df.iloc[:,0:4], df.iloc[:,-1], test_size=0.2, random_state=True)
```

```python
xtrain.head()
```

```python
xtest.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe= OneHotEncoder(drop='first', sparse_output=False)
```

here drop =first will remove the first column to avoid multicollinearity

```python
xtrain_new = ohe.fit_transform(xtrain[['fuel', 'owner']])
```

```python
xtest_new= ohe.transform(xtest[['fuel','owner']])
```

```python
xtrain_new.shape
```

```python
np.hstack((xtrain[['brand','km_driven']].values,xtrain_new)).shape
```

4. one hot encoding with top categories

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold=100
```

```python
repl = counts[counts <= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(repl,'uncommon')).sample(5)
```
