```python
import numpy as np
import pandas as pd
```

```python
df= pd.read_csv("/customer.csv")
```

```python
df.sample(3)
```

review and education is the ordinal categorical data column. Gender, review, education  are categorical.

Review and education - ordinal encoding
Purchase- label encoding

```python
df= df.iloc[:,2:]
```

```python
df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.model_selection import train_test_split
```

firstly, split train test

```python
x_train, x_test, y_train, y_test= train_test_split(df.iloc[:,0:2], df.iloc[:,-1], test_size=0.2)
```

```python
x_train
```

```python
oe= OrdinalEncoder(categories=[['Poor','Average','Good'], ['School', 'UG', 'PG']])
```

make sure to pass the values inside list according to importance . good and pg have more weightage

FIT FIRST THEN TRANSFORM

```python
oe.fit(x_train)
```

```python
x_train= oe.transform(x_train)
x_test= oe.transform(x_test)
```

```python
x_train
```

it has transformed the columns and given them values

```python
oe.categories
```
